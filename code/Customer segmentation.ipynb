```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
df=pd.read_csv('marketing_campaign.csv',sep='\t')
df.shape
```

```python
df.head()
```

```python
df.info()
#Data types of columns and missing values in income is observable
```

```python
print(df['Marital_Status'].unique())
print(df['Education'].unique())
# here 2n cycle and master are same level of eduaction
# replacing the 2ncycle to master 
df['Education']=df['Education'].replace('2n Cycle','Master')
df['Education'].unique()
```

```python
labelE=LabelEncoder()
df['Qualification']=labelE.fit_transform(df['Education'])

relationship_dummy=pd.get_dummies(df['Marital_Status'],prefix='m_status')
df=pd.concat([df,relationship_dummy],axis=1)
df.drop(columns='Marital_Status')
```

```python
df.info()
```

```python
df['Dt_Customer']=pd.to_datetime(df['Dt_Customer'],format='mixed')
df['Dt_Customer'].dtype
df=df.astype({'Education':'category','Marital_Status':'category'})
df.dtypes
```

```python
df['Income']=df['Income'].fillna(df['Income'].mean())
df['Income'].isna().sum()
```

```python
df['Total_amount_spent']=df['MntWines']+df['MntFruits']+df['MntMeatProducts']+df['MntFishProducts']+df['MntSweetProducts']+df['MntGoldProds']
df=df.drop(columns=['MntWines','MntFruits','MntMeatProducts','MntFishProducts','MntSweetProducts','MntGoldProds'])
df['Total_children']=df['Kidhome']+df['Teenhome']
df=df.drop(columns=['Kidhome','Teenhome'])
df['Total_interaction']=df['NumDealsPurchases']+df['NumWebPurchases']+df['NumCatalogPurchases']+df['NumStorePurchases']+df['NumWebVisitsMonth']
df=df.drop(columns=['NumDealsPurchases','NumWebPurchases','NumCatalogPurchases','NumStorePurchases','NumWebVisitsMonth'])
df['Total_campaign_accepted']=df['AcceptedCmp1']+df['AcceptedCmp2']+df['AcceptedCmp3']+df['AcceptedCmp4']+df['AcceptedCmp5']
df=df.drop(columns=['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5'])
df
```

```python
# Select numeric columns (exclude ID, Z_CostContact, Z_Revenue as they're not useful)
cols_to_plot = ['Year_Birth', 'Income', 'Total_children', 'Recency', 'Total_amount_spent','Total_interaction']

fig, axes = plt.subplots(3, 2, figsize=(12,8))
axes = axes.flatten()

for i, col in enumerate(cols_to_plot):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
#removing outlier from the important columns
birth_outliers_index=df.loc[df['Year_Birth']<1940].index
# print(birth_outliers_index)

child_outlier_index=df.loc[df['Total_children']>2].index
# print(child_outlier_index)

total_amount_spent=df.loc[df['Total_amount_spent']>2500].index
# print(total_amount_spent)

total_interaction=df.loc[df['Total_interaction']>40].index
# df=df.drop(total_interaction)

income_outliers_index=df.loc[df['Income']>115000].index
# print(income_outliers_index)

df=df.drop(income_outliers_index)
df=df.drop(total_amount_spent)
df=df.drop(total_interaction)
df=df.drop(child_outlier_index)
df=df.drop(birth_outliers_index)
```

```python
cols_to_plot = ['Year_Birth', 'Income', 'Total_children', 'Recency', 'Total_amount_spent','Total_interaction']

fig, axes = plt.subplots(3, 2, figsize=(12,8))
axes = axes.flatten()

for i, col in enumerate(cols_to_plot):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
key_features = ['Income', 'Total_amount_spent','Total_campaign_accepted', 'Total_interaction']
features = df[key_features]

SS = StandardScaler()
scaled_df = SS.fit_transform(features)
```

```python
pca = PCA(n_components=2)
pca_features = pca.fit_transform(scaled_df)

print(f"Explained variance: {sum(pca.explained_variance_ratio_):.2%}\n")

# Check silhouette scores
for n in [2,3,4,5]:
    model = KMeans(n_clusters=n, random_state=42)
    labels = model.fit_predict(pca_features)
    score = silhouette_score(pca_features, labels)
    print(f"K={n}, Silhouette Score: {score:.4f}")  
```

```python
model = KMeans(n_clusters=3, random_state=42)
labels = model.fit_predict(pca_features)
score = silhouette_score(pca_features, labels)
print(f"K=3, Silhouette Score: {score:.4f}")  

plt.figure(figsize=(10, 6))

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  

for cluster in [0, 1, 2]:
    mask = labels == cluster
    plt.scatter(pca_features[mask, 0], pca_features[mask, 1], 
                c=colors[cluster], label=f'Cluster {cluster}', alpha=0.6, s=50)

# Add centroids
centers = model.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='X', s=200, edgecolors='black', label='Centroids')

plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.title('Customer Segments (K=3)')
plt.legend()  
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans

# 1. We recreate your final K=3 model to get the exact labels
final_kmeans = KMeans(n_clusters=3, random_state=42)
final_labels = final_kmeans.fit_predict(pca_features)

# 2. We put your 2D PCA dots and their labels into a simple table
pca_df = pd.DataFrame(pca_features, columns=['PCA1', 'PCA2'])
pca_df['Cluster'] = final_labels

# 3. We save it as a CSV file
pca_df.to_csv('pca_clusters_background.csv', index=False)
print("Background data saved successfully!")
```
